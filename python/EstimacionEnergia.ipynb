# Primera estimación de energía

Semana 5 - "Modelo base validado".

Calcula el calor latente MÍNIMO (piso termodinámico) necesario para
evaporar el agua removida en cada una de las tres estrategias, usando
`estimacion_energia.py`. **No** es el consumo energético real (no
incluye electricidad de ventiladores, pérdidas térmicas, ni el efecto
de sorción del agua ligada) — ver el docstring del módulo para el
detalle completo de los supuestos y limitaciones.

```python
import sys
sys.path.append(".")

import pandas as pd
from estimacion_energia import estimar_todas, calor_latente_kj_kg
```

## Resultados para dos tamaños de lote

1000 kg (base unitaria, por tonelada) y 2500 kg (lote industrial
estándar, ~200 @).

```python
filas = []
for masa in (1000.0, 2500.0):
    for r in estimar_todas(masa):
        filas.append({
            "Lote (kg)": masa,
            "Estrategia": r.estrategia,
            "Agua removida (kg)": round(r.agua_removida_kg, 1),
            "T proceso (C)": r.temperatura_proceso_c,
            "L (kJ/kg)": round(r.calor_latente_kj_kg, 1),
            "Energia (kWh)": round(r.energia_kwh, 1),
        })

tabla = pd.DataFrame(filas)
tabla
```

## Lectura de los resultados

Las tres estrategias dan energías del mismo orden de magnitud
(300-450 kWh por tonelada), lo cual es esperable: el calor de
vaporización del agua cambia poco con la temperatura de proceso
(2360-2440 kJ/kg en el rango de interés). La diferencia principal entre
estrategias viene de cuánta agua hay que remover según el M0/Mf que
reporta cada fuente bibliográfica (no equalizados entre sí en este
v0.1) — no se debe leer todavía como una comparación de eficiencia
energética real entre métodos; eso requiere fijar un M0/Mf común y
medir/estimar el consumo eléctrico real (semana 9).
